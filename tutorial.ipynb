In PyTorch, `torch.random` is not a function by itself, but `torch` does provide a module for random number generation, which is used for various purposes, such as sampling random numbers or initializing tensors.

Here are some common random functions available in PyTorch:

### 1. **`torch.rand()`**
Generates a tensor with random numbers drawn from a uniform distribution between 0 and 1.

```python
import torch

# Create a 3x3 tensor of random numbers between 0 and 1
tensor = torch.rand(3, 3)
print(tensor)
```

### 2. **`torch.randn()`**
Generates a tensor with random numbers drawn from a normal (Gaussian) distribution with mean 0 and variance 1.

```python
# Create a 3x3 tensor of random numbers from a normal distribution
tensor = torch.randn(3, 3)
print(tensor)
```

### 3. **`torch.randint()`**
Generates a tensor with random integers from a specified low (inclusive) to high (exclusive) range.

```python
# Create a 3x3 tensor of random integers between 0 and 10
tensor = torch.randint(0, 10, (3, 3))
print(tensor)
```

### 4. **`torch.rand_like()`**
Generates a random tensor with the same size as another tensor, using a uniform distribution.

```python
# Create a tensor and generate a random tensor of the same shape
tensor = torch.randn(3, 3)
random_tensor = torch.rand_like(tensor)
print(random_tensor)
```

### 5. **`torch.randint_like()`**
Generates a random tensor of integers with the same shape as another tensor.

```python
# Create a tensor and generate a tensor of random integers of the same shape
tensor = torch.randn(3, 3)
random_tensor = torch.randint_like(tensor, 0, 10)
print(random_tensor)
```

### 6. **`torch.manual_seed()`**
Sets the seed for generating random numbers, ensuring reproducibility.

```python
torch.manual_seed(42)  # Set a seed value for reproducibility
tensor = torch.rand(3, 3)
print(tensor)
```

# BASIC NEURAL NETWORK

Awesome! Let's **visualize** how data flows through a `nn.Linear` layer and see how the **weights** are structured.

---

## 🏗️ **`nn.Linear(10, 5)` Example (Detailed Breakdown)**

- **Inputs**: `10` features (think of it as 10 numbers coming in)
- **Outputs**: `5` features (the layer gives 5 numbers out)
  
Each **output neuron** is connected to **every input neuron**, and each connection has a **weight**.

---

### ✅ **How the Weight Matrix Works**

- The **weight matrix** in `nn.Linear(10, 5)` has:
  - **5 rows** (one for each output neuron)
  - **10 columns** (one for each input neuron)
  
➡️ Shape of the weight matrix: `[5, 10]`

You can think of it like this:
```
        input features
        1   2   3   ...  10
     +-------------------------
 o1  | w11 w12 w13 ... w1,10   ← weights for output 1
 o2  | w21 w22 w23 ... w2,10   ← weights for output 2
 o3  | w31 w32 w33 ... w3,10
 o4  | w41 w42 w43 ... w4,10
 o5  | w51 w52 w53 ... w5,10
```
- Each **row** computes one **output feature**.
- Each **column** represents weights connected to **one input feature**.

---

### ✅ **How Data Flows Through the Layer**

#### 1. You give an **input vector** of shape `[10]`, e.g.:
```
x = [x1, x2, x3, ..., x10]
```

#### 2. It **multiplies** the input vector by the **transpose** of the weight matrix (`W.T`), and **adds the bias**:
```
output = x @ W.T + bias
```
➡️ Output shape: `[5]`  
➡️ You get one number for each output neuron.

---

### ✅ **Concrete Example (with numbers)**

#### Inputs:
```
x = [1.0, 2.0, 3.0, ..., 10.0]  # (10 values)
```

#### Randomly initialized weights (simplified for 3 inputs/2 outputs to keep it small):
```
W = [[0.1, 0.2, 0.3],   # weights for output neuron 1
     [0.4, 0.5, 0.6]]   # weights for output neuron 2
```
#### Bias:
```
b = [0.1, -0.1]
```

#### Calculation:
```
Output neuron 1:
(1.0 * 0.1) + (2.0 * 0.2) + (3.0 * 0.3) + bias 0.1
= 0.1 + 0.4 + 0.9 + 0.1 = 1.5

Output neuron 2:
(1.0 * 0.4) + (2.0 * 0.5) + (3.0 * 0.6) - 0.1
= 0.4 + 1.0 + 1.8 - 0.1 = 3.1
```

➡️ Final output:  
```
[1.5, 3.1]
```

---

### ✅ **Visual Data Flow**
```
         Inputs (10 features)
        ┌────┬────┬────┬────┬────┐
        │ x1 │ x2 │ x3 │... │ x10│
        └─┬──┴─┬──┴─┬──┴────┴────┘
          │     │     │
          ▼     ▼     ▼
       [weights] [weights] ...  → Multiply each by weights → Sum + bias → Output neuron 1
       [weights] [weights] ...  → Multiply each by weights → Sum + bias → Output neuron 2
          ...
       Repeat for all 5 outputs
```

---

### TL;DR:
- A `nn.Linear(10, 5)` layer takes **10 numbers**, combines them using **weights and biases**, and gives **5 output numbers**.
- **Weights** control the **influence** of each input on each output.
- The **forward pass** is just a **matrix multiplication** plus **bias**.

### ✅ **Activation Functions and Where They're Applied**

In a neural network with:
1. **1 Input Layer**
2. **1 Hidden Layer**
3. **1 Output Layer**

You typically do something like this:
```
Input Layer → Linear Layer → Activation Function → Linear Layer → (maybe another Activation)
```

---

### 🔧 **Where Activation Functions Are Applied**
- **After each Linear Layer**, you usually apply an **activation function** (like ReLU, Sigmoid, Softmax).
  
#### Here's the flow:
1. **Input Layer**  
   - Just your raw data/features.  
   - No computation, no activation.
   
2. **Hidden Layer**  
   - You apply a **Linear transformation** (`nn.Linear`), which gives you **weighted sums + bias**.  
   - After this, you usually apply an **activation function** like **ReLU** or **Sigmoid** to introduce **non-linearity**.

3. **Output Layer**  
   - Another **Linear transformation**.  
   - Then you **may** apply an activation function depending on:
     - **Regression**: Usually no activation or sometimes **ReLU**.
     - **Binary classification**: Often use **Sigmoid**.
     - **Multiclass classification**: Often use **Softmax**.

---

### ✅ **Why Apply Activation Functions?**
- **Linear layers alone** just apply a **linear transformation** (matrix multiplication + bias).
- If you **stack only linear layers**, the entire network can still be simplified into **one big linear equation**, which **limits** what it can learn.
- **Activation functions** introduce **non-linearity**, allowing the network to **learn complex patterns**.

---

### ✅ **Example in Code:**
```python
import torch.nn as nn

class SimpleNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(10, 5)  # First linear layer
        self.relu = nn.ReLU()        # Activation function
        self.fc2 = nn.Linear(5, 1)   # Output linear layer
        self.sigmoid = nn.Sigmoid()  # Output activation (for binary classification)

    def forward(self, x):
        x = self.fc1(x)      # Linear transformation
        x = self.relu(x)     # Activation function applied after first layer
        x = self.fc2(x)      # Another linear transformation
        x = self.sigmoid(x)  # Activation for the output
        return x
```

---

### ✅ **Summary (For 1 Input, 1 Hidden, 1 Output Layer)**
1. **Inputs** → no activation applied to raw data.
2. **After Hidden Layer Linear transformation**, apply an **activation function** (ReLU/Sigmoid/Tanh).
3. **After Output Layer**, **maybe** apply another activation depending on the task.

````python
import torch
import torch.nn as nn
import torch.nn.functional as F
````

Linear - subclass of Module
Module a class inside torch.nn
nn-a sub module of torch

````python
# Create a model class that inherits nn.module
class myModel(nn.Module):
 #using the iris dataset -4 features for input layer -> hidden layer1 (h1) ->hidden layer h2->output (3 classes of iris flowers)
 def __init__(self,in_features=4,h1=8,h2=9, out_features=3):
  super().__init__()  # instantiate nn.Module
  self.fc1 =nn.Linear(in_features,h1) #linear transformation is applied the in_features=no. of features from input layer to hidden layer 1
  # linear transformation -> y = xW.T + b
  self.fc2=nn.Linear(h1,h2) # linear transformation from h1 to h2
  self.out=nn.Linear(h2,out_features)  #linear transformation from h2 to output layer

 def forward(self,x):
  x=F.relu(self.fc1(x))
  x=F.relu(self.fc2(x))
  x=self.out(x)

  return x
````

````python
#pick a random seed for rantorch.domization
torch.manual_seed(32)
#Create an instance of  the model
model=myModel()
print(model)
````

````python
my_df['species']=my_df['species'].replace('setosa',0.0)  #replacing the name with a number -for machine learning - data as number
my_df['species']=my_df['species'].replace('versicolor',1.0)
my_df['species']=my_df['species'].replace('virginica',2.0)
my_df.head(10)
````

````python
# Train Test split set X,y
X=my_df.drop('species',axis=1)  # dropping the last column - which is the output
y=my_df['species']

#convert these into numpy arrays
X=X.values
y=y.values
````

````python
from sklearn.model_selection import train_test_split
````

If you do set random_state (like random_state=41), the split will be the same every time you run it.
Setting random_state ensures that you (and others) always get the same training and testing data. This makes debugging, sharing, and comparing models consistent.

````python
# Train test split ratio= 80:20
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=32)
````

````python
#converting it all to tensor for putorch
# FloatTensor bcz all the feature values are float if int - Tensor should do
X_train=torch.FloatTensor(X_train)
X_test=torch.FloatTensor(X_test)
# y values are converted to long as we dont necessarily need the output as decimal - we only care if they are 0/1/2
y_train=torch.LongTensor(y_train)
y_test=torch.LongTensor(y_test)
````

````python
#set the criterion of model to measure the error , how far off the predictions ar from the data
criterion=nn.CrossEntropyLoss()
#sets the loss function (or criterion) for your model. This is how the model measures how far off its predictions are from the actual labels, during training.
#crosss entropy loss is generally applied for multi class classification problems

#chose optimizer, and learning rate
optimizer = torch.optim.Adam(model.parameters(),lr=0.001)  #here we can add additional parameters like weight decay
````

If error doesnt go down after bunch of iterations (epochs) lower the lr

The model.parameters point to the model parameters (fc1,fc2,out)

Optimizer -Adam optimizer - a type of optimization method - like AdaGrad, Gradient descent etc.

One epoch - one run through our entire data


````python
#Train our model
epoch = 300
losses=[] # for entering losses - to know if our model is learning
for i in range(epoch):
  y_pred = model.forward(X_train) #get predicted results
  loss= criterion(y_pred,y_train) #measure the oss against our original value
  losses.append(loss.detach().numpy())

  if i%10 ==0:
    print(f'epoch: {i} and loss: {loss}')
  #do some back propogation
  optimizer.zero_grad()
  loss.backward()
  optimizer.step()
````

Great question! Let's break this down clearly:

---

### ✅ `optimizer.zero_grad()`
- **What it does**: Resets the **gradients** of all the model parameters to zero.
- **Why?** Gradients accumulate by default in PyTorch. If you don't clear them, they add up across batches/iterations, which can mess up your training.
  
  > Think of it like wiping the slate clean before writing new gradient values.

---

### ✅ `loss.backward()`
- **What it does**: Computes the **gradients** of the **loss** with respect to **model parameters**.
- **Why?** This is **backpropagation**. It figures out how much each parameter contributed to the error (loss), so we know how to adjust them.
  
  > PyTorch calculates these gradients and stores them in the `.grad` attribute of each parameter.

---

### ✅ `optimizer.step()`
- **What it does**: Updates the **model parameters** (weights and biases) using the gradients computed in `loss.backward()` and according to the **optimizer** (Adam, SGD, etc.).
- **Why?** This moves the parameters in a direction that should **reduce the loss**—i.e., it improves the model.

---

### ✅ Simple Analogy:
Imagine training like hiking downhill in fog:
1. `loss.backward()` → You check which direction is downhill (gradients).
2. `optimizer.step()` → You take a step in that direction (update parameters).
3. `optimizer.zero_grad()` → You forget the last direction so you can calculate a new one (clear gradients for the next step).

---

### ✅ Full cycle in training:
```python
optimizer.zero_grad()  # Reset gradients to zero
outputs = model(X_batch)  # Forward pass
loss = criterion(outputs, y_batch)  # Calculate loss
loss.backward()  # Compute gradients (backpropagation)
optimizer.step()  # Update weights (parameters)
```
